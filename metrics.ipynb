```python
from zss import simple_distance, Node
import json
import os
```

```python
folder_name_1 = "2023년도_지역특화사업_이야기_-_통영_한산마을_~_함안_여항면_다랑논_~_산청_단계마을(Page47)"
folder_name_2 = "Let's_Learn_Korean(Page6)"
```

```python

symbolic_tree_1 = os.path.join("datasets", folder_name_1, "symbolic_tree.json")
symbolic_tree_2 = os.path.join("datasets", folder_name_2, "symbolic_tree.json")

with open(symbolic_tree_1, "r", encoding="utf-8") as file:
        json_data_1 = json.load(file)

with open(symbolic_tree_2, "r", encoding="utf-8") as file:
        json_data_2 = json.load(file)
```

```python
def json_to_tree(data):
    if isinstance(data, dict):
        node = Node(data.get("symbol_name", "root"))
        if "children" in data:
            for child in data["children"]:
                node.addkid(json_to_tree(child))
        return node
    elif isinstance(data, str):
        return Node(data)
    else:
        raise ValueError("Unsupported JSON format")
```

```python
tree_a = json_to_tree(json_data_1)
tree_b = json_to_tree(json_data_2)
```

```python
distance = simple_distance(tree_a, tree_b)
distance
```
